```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
from skimage import io
from skimage.feature import blob_log
```

```python
fname = "/Users/sean/Desktop/Repos/LAB-positioner-tilt/data/Repeatability_results_100.npy"
```

```python
foundSpots = np.load(fname)
```

```python
baseDir = "/Users/sean/Desktop/Repos/ThorlabsImages/imgs/lightsOff_LEDOn"
theta=np.arange(0,2*np.pi,step=0.01)
allFiles = os.listdir(baseDir)

for f,blobs in zip(allFiles,foundSpots):
    tifFile = os.path.join(baseDir,f)
    img = io.imread(tifFile, as_gray=True)
    fig,ax = plt.subplots()
    ax.imshow(img,vmin=7,vmax=20)
    for blob in blobs:
        ax.scatter(blob[1]-blob[2]*np.cos(theta),blob[0]-blob[2]*np.sin(theta),ls='--')
        ax.scatter(blob[1],blob[0],marker='x',color='red')
    plt.show()
    del fig,ax
```
